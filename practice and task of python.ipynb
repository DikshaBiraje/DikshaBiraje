```python
#11.Reverse a number. 
num = int(input("Enter a number: "))
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

print("Reverse =", rev)
```

```python
#12.Check whether a number is palindrome. 
num = int(input("Enter a number: "))
temp = num
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

if temp == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#13.Check whether a number is Strong Number. 
num = int(input("Enter a number: "))
temp = num
sum = 0
digits = len(str(num))

while temp > 0:
    digit = temp % 10
    sum += digit ** digits
    temp = temp // 10

if sum == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
#14.Check whether a number is Perfect Number. 
num = int(input("Enter a number: "))
temp = num
sum = 0

while temp > 0:
    digit = temp % 10
    fact = 1

    for i in range(1, digit + 1):
        fact *= i

    sum += fact
    temp = temp // 10

if sum == num:
    print("Strong Number")
else:
    print("Not a Strong Number")
```

```python
#15.Check whether a number is Neon Number. 
num = int(input("Enter a number: "))
sum = 0

for i in range(1, num):
    if num % i == 0:
        sum += i

if sum == num:
    print("Perfect Number")
else:
    print("Not a Perfect Number")
```

```python
#16.Check whether a number is Spy Number.
num = int(input("Enter a number: "))
square = num * num
sum = 0

while square > 0:
    digit = square % 10
    sum += digit
    square = square // 10

if sum == num:
    print("Neon Number")
else:
    print("Not a Neon Number")
```

```python
#17.Check whether a number is Automorphic Number. 
n = int(input("Enter a number: "))
temp = n
sum_digits = 0
product_digits = 1

while temp > 0:
    digit = temp % 10
    sum_digits += digit
    product_digits *= digit
    temp //= 10

if sum_digits == product_digits:
    print("Spy Number")
else:
    print("Not a Spy Number")
```

```python
#18.Check whether a number is Harshad (Niven) Number. 
n = int(input("Enter a number: "))
square = n * n

if str(square).endswith(str(n)):
    print("Automorphic Number")
else:
    print("Not an Automorphic Number")
```
