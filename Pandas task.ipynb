<h2><b>4. Combining DataFrames</b></h2>

```python
# 1.# Combine two DataFrames by adding rows
df1 = pd.DataFrame({
    "Name": ["Amit", "Priya"],
    "Age": [20, 21]
})
df2 = pd.DataFrame({
    "Name": ["Rahul", "Sneha"],
    "Age": [19, 22]
})
# Combine rows
result = pd.concat([df1, df2])
print(result)
```

```python
# 2. Combine two DataFrames by adding columns
df1 = pd.DataFrame({
    "Name": ["Amit", "Priya"]
})
df2 = pd.DataFrame({
    "Marks": [85, 90]
})
# Combine columns
result = pd.concat([df1, df2], axis=1)
print(result)
```

```python
# 3. Merge two DataFrames using the common column "ID"
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Amit", "Priya", "Rahul"]
})
df2 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 90, 78]
})
# Merge DataFrames on ID column
result = pd.merge(df1, df2, on="ID")
print(result)
```

```python
# 4. Join two DataFrames using their index
df1 = pd.DataFrame({
    "Name": ["Amit", "Priya"]
}, index=[1, 2])
df2 = pd.DataFrame({
    "Marks": [85, 90]
}, index=[1, 2])
# Join DataFrames based on index
result = df1.join(df2)
print(result)
```

```python
# Add one DataFrame below another and reset the index
df1 = pd.DataFrame({
    "Name": ["Amit"],
    "Age": [20]
})
df2 = pd.DataFrame({
    "Name": ["Priya"],
    "Age": [21]
})
# Append rows and create a new index
result = pd.concat([df1, df2], ignore_index=True)
print(result)
```

<h2><b>9. Map, Apply, DataFrame.map, And Lambda Function</b></h2>

```python
# 1. Convert Gender values into numeric codes using map()
df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Sneha"],
    "Gender": ["Male", "Female", "Male", "Female"]
})
gender_code = {
    "Male": 1,
    "Female": 0
}
df["Gender_Code"] = df["Gender"].map(gender_code)
print(df)
```

```python
#2. Convert names to uppercase using map()
df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul"]
})
df["Upper_Name"] = df["Name"].map(str.upper)
print(df)
```

```python
# Add 5 bonus marks using apply()
df = pd.DataFrame({
    "Marks": [75, 82, 90]
})
df["Updated_Marks"] = df["Marks"].apply(lambda x: x + 5)
print(df)
```

```python
# 4. Find total marks using apply()
df = pd.DataFrame({
    "Math": [80, 70, 90],
    "Science": [85, 75, 95]
})
df["Total"] = df[["Math", "Science"]].apply(sum, axis=1)
print(df)
```

```python
# 6. Multiply every value by 10 using DataFrame.map()
df = pd.DataFrame({
    "A": [1, 2],
    "B": [3, 4]
})
result = df.map(lambda x: x * 10)
print(result)
```

```python
# 7. Check whether marks are Pass or Fail
df = pd.DataFrame({
    "Marks": [35, 80, 25, 60]
})
df["Result"] = df["Marks"].apply(lambda x: "Pass" if x >= 35 else "Fail")
print(df)
```

```python
# 8. Convert student names to uppercase
df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul"]
})
df["Upper_Name"] = df["Name"].apply(lambda x: x.upper())
print(df)
```

<h2><b>14. GroupBy: Split, Apply, Combine</b></h2>

```python
# 1.Split data into groups based on Department
df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Sneha"],
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [50000, 40000, 55000, 45000]
})
group = df.groupby("Department")
print(df)
```

```python
# 2. Count the number of employees in each department
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR", "IT"]
})
print(df.groupby("Department").size())
```

```python
# 3. Display only the IT department data
df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul"],
    "Department": ["IT", "HR", "IT"]
})
group = df.groupby("Department")
print(group.get_group("IT"))
```

```python
# 4. Calculate the average salary for each department
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [50000, 40000, 55000, 45000]
})
print(df.groupby("Department").agg({"Salary": "mean"}))
```

```python
# 5. Add department-wise average salary to each employee
df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [50000, 40000, 55000, 45000]
})
df["Avg_Salary"] = df.groupby("Department")["Salary"].transform("mean")
print(df)
```
